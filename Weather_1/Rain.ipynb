```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv('BOM_Rain.csv')
df.head()
```

```python
df.describe()
```

```python
plt.figure(figsize=(12, 4))
plt.plot(df['Year'], df['Total'], linestyle='-', color='r')
plt.title('Total Annual Rainfall in Mumbai (121 years)')
plt.xlabel('Year')
plt.ylabel('Total Rainfall (mm)')
plt.grid(True)
plt.show()
```

```python
monthly_mean = df[['Jan', 'Feb', 'Mar', 'April', 'May', 'June', 'July', 'Aug', 'Sept', 'Oct', 'Nov', 'Dec']].mean()

plt.figure(figsize=(12, 4))
plt.plot(monthly_mean.index, monthly_mean.values, marker='o', linestyle='-', color='b')

plt.title('Average Monthly Rainfall in Mumbai (121 years)')
plt.xlabel('Month')
plt.ylabel('Average Rainfall (mm)')
plt.grid(True)

plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'April', 'May', 'June', 'July', 'Aug', 'Sept', 'Oct', 'Nov', 'Dec']

for month in months:
    plt.figure(figsize=(12,2))
    plt.plot(df['Year'], df[month], linestyle='-', color='b')
    plt.title(f'Average Rainfall in Mumbai - {month}')
    plt.xlabel('Year')
    plt.ylabel('Rainfall (mm)')
    plt.grid(True)
    plt.show()
```
