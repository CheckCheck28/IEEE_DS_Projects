```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv('T2M_all_seasons.csv')
```

```python
df.head()
df.describe()
```

```python
seasons = ['Winter', 'PreMonsoon', 'Monsoon', 'PostMonsoon']
seasonal_colors = ['b', 'g', 'r', 'm']
plt.figure(figsize=(12, 4))
for i, season in enumerate(seasons):
        plt.plot(df['Year'], df[season],linestyle='-', color = seasonal_colors[i])
plt.title('Average Temperature in Kolkata')
plt.xlabel('Year')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.legend(seasons)
plt.show()
```

```python
months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']
for month in months:
    plt.figure(figsize=(12,2))
    plt.plot(df['Year'], df[month], linestyle='-', color='b')
    plt.title(f'Average Temperature in Kolkata - {month}')
    plt.xlabel('Year')
    plt.ylabel('Temperature (°C)') 
    plt.grid(True)
    plt.show()


```
